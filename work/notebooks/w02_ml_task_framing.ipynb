## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

The pattern is too messy for an if-statement because multiple features interact in complex and non-linear ways. For example, a page may have high impressions but low CTR, or strong engagement but a poor average position, and deciding which one matters more cannot be captured with simple rules. Additionally, these relationships vary across different content types and change over time, making fixed thresholds unreliable. Machine learning models can automatically learn these hidden patterns and trade-offs, while an if-statement would be too rigid and fail to generalize across different cases.
